```python
import os
import time
import sys

# Get project root (go up from notebooks/)
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))

# Add /src to Python path
sys.path.append(os.path.join(PROJECT_ROOT, "src"))

from config import DATA_PATH, OUTPUT_DIR
from data_preprocessing import run_preprocessing, print_summary
from eda import run_eda
from rfm_analysis import run_rfm
from k_means import run_k_means
from export import run_export

def run_pipeline(data_path=DATA_PATH, output_dir=OUTPUT_DIR):
    t_start = time.time()

    os.makedirs(output_dir, exist_ok=True)

    print("\n" + "═" * 60)
    print("  ONLINE RETAIL – CUSTOMER ANALYSIS PIPELINE")
    print("═" * 60)

    print("\n[STEP 1/5] Data Loading & Preprocessing")
    df = run_preprocessing(data_path)
    print_summary(df)

    print("\n[STEP 2/5] Exploratory Data Analysis")
    metrics = run_eda(df, output_dir)

    print("\n[STEP 3/5] RFM Analysis")
    rfm_df = run_rfm(df, output_dir)

    print("\n[STEP 4/5] Customer Segmentation")
    rfm_segmented = run_k_means(rfm_df, output_dir)

    print("\n[STEP 5/5] Exporting Results")
    exported_paths = run_export(rfm_segmented, output_dir)

    elapsed = time.time() - t_start

    print("\n" + "═" * 60)
    print(f"  PIPELINE COMPLETE  ({elapsed:.1f}s)")
    print("═" * 60)
    print(f"  Outputs saved to: {output_dir}\n")

    for label, path in exported_paths.items():
        print(f"  {label:<20} → {os.path.basename(path)}")

    return {
        "df": df,
        "rfm": rfm_df,
        "rfm_segmented": rfm_segmented,
        "metrics": metrics,
    }
```

```python
results = run_pipeline()
```
